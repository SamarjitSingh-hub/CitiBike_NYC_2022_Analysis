```python
weather_data = json.loads(response.text)
```

```python
# 5. Extract Dates and Temperatures
avg_temps = [item for item in weather_data['results'] if item['datatype'] == 'TAVG']
```

```python
dates_temp = [item['date'] for item in avg_temps]
```

```python
temps = [item['value'] for item in avg_temps]
```

```python
# 6. Create Weather DataFrame and Export to CSV
df_temp = pd.DataFrame()
```

```python
df_temp['date'] = [datetime.strptime(d, "%Y-%m-%dT%H:%M:%S") for d in dates_temp]
```

```python
df_temp['avgTemp'] = [float(v) / 10.0 for v in temps]
```

```python
df_temp.to_csv('Weather_Data_2022.csv', index=False) # Export to CSV
```

```python
# Prepare for merger
df['started_at'] = pd.to_datetime(df['started_at'])
```

```python
df['date'] = df['started_at'].dt.date
```

```python
# Fix data types before merging( Dtype should both to same being "datetime")
df['date'] = pd.to_datetime(df['date'])
df_temp['date'] = pd.to_datetime(df_temp['date'])
```

```python
# 8. Merge CitiBike Trip Data with Weather Data
df_merged = df.merge(df_temp, how='left', on='date', indicator=True)
```
